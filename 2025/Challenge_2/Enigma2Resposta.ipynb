Encontre aquele que teve mais acessos : Nesse caso, Vampire Squid (Esse é o codenome do hacker)

```python
out.groupby(
    'quem-acessou'
).agg({'quantidade-de-acessos' : sum}).sort_values(by='quantidade-de-acessos', ascending=False)
```
